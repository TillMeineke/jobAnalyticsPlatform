```python
import requests

role = "data-scientist"
location = "hamburg"

url = f"https://www.stepstone.de/jobs/{role}/in-{location}?radius=30"

# Simulate a real browser request
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36",
    "Accept-Language": "en-US,en;q=0.9",
    "Referer": "https://www.google.com/",
}

# Make the request
response = requests.get(url, headers=headers, timeout=10)
# response = urlopen(url, headers=headers, timeout=10)

# Check response
if response.status_code == 200:
    print("Page loaded successfully!")
    print(response.text[:500])  # Print first 500 characters
else:
    print(f"Failed with status: {response.status_code}")
```

```python
soup = BeautifulSoup(response.text, "html.parser")
type(soup)
```

```python
related_roles = soup.find_all("a", attrs={"data-at": "related-jobs-link"})

unique_roles = set()

new_related_roles = []

for role in related_roles:
    unique_roles.add(role.text.strip())

for role in unique_roles:
    new_related_roles.append(role)
```

```python
jobs = soup.find_all("a", attrs={"data-at": "job-item-title"})
print(jobs[0].prettify())
```

```python
job = jobs[0]

# Extract job title
title = job.get_text(strip=True)

# Extract job URL (append the base URL if necessary)
link = job["href"]

# Extract data attributes
data_at = job.get("data-at")
data_testid = job.get("data-testid")

# Print extracted fields
print(f"Title: {title}")
print(f"Link: {link}")
print(f"Data-at: {data_at}")
print(f"Data-testid: {data_testid}")
```

```python
import pandas as pd

# Create a DataFrame
data = {
    "Name": ["John", "Alice", "Bob"],
    "Skills": ["Python, Java", "C++, Python", "Java"],
}
df = pd.DataFrame(data)
# Split the 'Skills' column into multiple rows
df = df.assign(Skills=df["Skills"].str.split(", ")).explode("Skills")
print(df)
```
